```python
def run_deutsch_jozsa_experiment(n=4):
    """Run the Deutsch-Jozsa experiment"""
    
    circuit, oracle_type, oracle_param = create_deutsch_jozsa_circuit(n)
    
    # Display the circuit
    print("\nCircuit:")
    print(circuit.draw())
    
    # Run on simulator
    simulator = AerSimulator()
    pass_manager = generate_preset_pass_manager(optimization_level=1, backend=simulator)
    transpiled_circuit = pass_manager.run(circuit)
    
    job = simulator.run(transpiled_circuit, shots=1024)
    result = job.result()
    counts = result.get_counts(transpiled_circuit)
    
    # Analyze results
    print("\nMeasurement results:")
    for bitstring, count in counts.items():
        print(f"Bitstring: {bitstring}, Count: {count}")
    
    # Check prediction
    all_zero_bitstring = '0' * n
    measured_constant = all_zero_bitstring in counts and len(counts) == 1
    
    print(f"\nAlgorithm prediction:")
    if measured_constant:
        print("The function is CONSTANT")
    else:
        print("The function is BALANCED")
    
    print(f"\nActual oracle type:")
    if oracle_type == 0:
        print(f"CONSTANT (value = {oracle_param})")
    else:
        print(f"BALANCED (parameter a = {oracle_param})")
    
    # Verify correctness
    correct_prediction = (measured_constant and oracle_type == 0) or (not measured_constant and oracle_type == 1)
    print(f"\nPrediction correct: {correct_prediction}")
    
    if not correct_prediction:
        print("\n⚠️  Something went wrong!")
    else:
        print("\n✅ Algorithm worked correctly!")
    
    # Plot results
    plot_histogram(counts)
    plt.title(f'Deutsch-Jozsa Results (n={n})')
    plt.show()
    
    return correct_prediction
```

```python
def test_multiple_runs(n=4, num_tests=5):
    """Test the algorithm multiple times"""
    print(f"Testing Deutsch-Jozsa Algorithm {num_tests} times with n={n}")
    print("=" * 60)
    
    success_count = 0
    
    for i in range(num_tests):
        print(f"\n--- Test {i+1} ---")
        success = run_deutsch_jozsa_experiment(n)
        if success:
            success_count += 1
    
    print(f"\n\nOverall Results:")
    print(f"Successful predictions: {success_count}/{num_tests}")
    print(f"Success rate: {success_count/num_tests*100:.1f}%")
```
